```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Count': df.isnull().sum(),
    'Missing Percentage': missing_percentage
})

display(missing_summary)
```

```python
categorical_columns = df.select_dtypes(include='object').columns

for column in categorical_columns:
    print(f"\nUnique values in {column}:")
    print(df[column].unique())
```

```python
for column in df.select_dtypes(include='object').columns:
    df[column] = df[column].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

print("Extra spaces removed from text columns.")
```

```python
missing_markers = [
    '', ' ', 'NA', 'N/A', 'na', 'n/a',
    'NULL', 'null', '-', '--'
]

df = df.replace(missing_markers, np.nan)

print("Missing-value markers converted to NaN.")
```

```python
print("Current data types:")
display(df.dtypes)
```

```python
# Convert commonly expected numeric columns if they exist

numeric_candidates = [
    'age',
    'salary',
    'experience',
    'years_of_experience',
    'performance_score',
    'department_id'
]

for column in numeric_candidates:
    if column in df.columns:
        df[column] = pd.to_numeric(df[column], errors='coerce')

print("Data types after conversion:")
display(df.dtypes)
```

```python
for column in df.columns:
    if 'date' in column.lower():
        df[column] = pd.to_datetime(df[column], errors='coerce')

print("Date columns converted where applicable.")
```

```python
for column in df.columns:
    if column in ['gender', 'sex']:
        df[column] = df[column].replace({
            'M': 'Male',
            'm': 'Male',
            'male': 'Male',
            'F': 'Female',
            'f': 'Female',
            'female': 'Female'
        })

print("Categorical values standardized where applicable.")
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate records:")
print(duplicate_count)
```

```python
print("Duplicate records:")
display(df[df.duplicated(keep=False)])
```

```python
rows_before_duplicates = len(df)

df = df.drop_duplicates()

rows_after_duplicates = len(df)

print("Rows before removing duplicates:", rows_before_duplicates)
print("Rows after removing duplicates:", rows_after_duplicates)
print("Duplicates removed:", rows_before_duplicates - rows_after_duplicates)
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

print("Missing values in numerical columns:")

for column in numeric_columns:
    missing = df[column].isnull().sum()
    if missing > 0:
        print(f"{column}: {missing}")
```

```python
for column in numeric_columns:
    if df[column].isnull().sum() > 0:
        median_value = df[column].median()
        df[column] = df[column].fillna(median_value)

print("Missing numerical values filled using median imputation.")
```

```python
categorical_columns = df.select_dtypes(include='object').columns

for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        mode_value = df[column].mode()[0]
        df[column] = df[column].fillna(mode_value)

print("Missing categorical values filled using mode imputation.")
```

```python
remaining_missing = df.isnull().sum().sum()

print("Missing values before forward fill:", remaining_missing)

if remaining_missing > 0:
    df = df.ffill()

print("Missing values after forward fill:", df.isnull().sum().sum())
```
